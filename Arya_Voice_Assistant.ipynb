```python
# Arya Voice Assistant Loop
while True:
    try:
        # Listen for the user's voice
        with srec.Microphone() as source:
            print("Arya is listening...")
            audio = recording.listen(source, timeout=10)
        print("Audio received!")

        # Convert speech to text
        command = recording.recognize_google(audio)
        command = command.lower()
        print("You said:", command)

        # Tokenizing the user's command
        tokens = wordpunct_tokenize(command)
        print("Tokens:", tokens)

        # Stemming the tokens
        stemmer = PorterStemmer()
        stemmed_tokens = [stemmer.stem(token) for token in tokens]
        print("Stemmed Tokens:", stemmed_tokens)

        # Identifying the user's intent
        if "your" in stemmed_tokens and "name" in stemmed_tokens:
            intent = "arya_name"
        elif (
            ("my" in stemmed_tokens and "name" in stemmed_tokens)
            or ("call" in stemmed_tokens and "me" in stemmed_tokens)
            ):
            intent = "user_name"
        elif "time" in stemmed_tokens:
            intent = "time"
        elif "date" in stemmed_tokens:
            intent = "date"
        elif "day" in stemmed_tokens and "date" not in stemmed_tokens:
            intent = "day"
        elif "hello" in stemmed_tokens or "hi" in stemmed_tokens or "hey" in stemmed_tokens:
            intent = "greeting"
        elif "how" in stemmed_tokens and "you" in stemmed_tokens:
            intent = "how_are_you"
        elif "goodby" in stemmed_tokens or "bye" in stemmed_tokens or "exit" in stemmed_tokens:
            intent = "goodbye"
        elif (
            ("not" in stemmed_tokens and "well" in stemmed_tokens)
            or ("not" in stemmed_tokens and "fine" in stemmed_tokens)
            or ("not" in stemmed_tokens and "good" in stemmed_tokens)
            or "sad" in stemmed_tokens
            or "tire" in stemmed_tokens
            ):
            intent = "user_feeling_bad"
        elif (
            "well" in stemmed_tokens
            or "good" in stemmed_tokens
            or "amazing" in stemmed_tokens
            or "fine" in stemmed_tokens
            ):
            intent = "user_doing_well"
        elif (
            ("i" in stemmed_tokens and "am" in stemmed_tokens)
            or ("i" in stemmed_tokens and "m" in stemmed_tokens)
            ):
            intent = "user_name"
        elif "thank" in stemmed_tokens:
            intent = "thanks"
        else:
            intent = "unknown"

        print("Intent is:", intent)

        # Responding based on the user's intent
        if intent == "arya_name":
            print("My name is Arya, and you?")

        elif intent == "user_name":
            if "my name is" in command:
                name = command.replace("my name is", "").strip()
            elif "call me" in command:
                name = command.replace("call me", "").strip()
            elif "i am" in command:
                name = command.replace("i am", "").strip()
            elif "i'm" in command:
                name = command.replace("i'm", "").strip()
            print("Nice to meet you,", name.title())

        elif intent == "how_are_you":
            print("I am doing very well, and you?")

        elif intent == "time":
            current_time = datetime.datetime.now().strftime("%I:%M %p")
            print("The current time is", current_time)

        elif intent == "date":
            current_date = datetime.datetime.now().strftime("%B %d, %Y")
            print("Today's date is", current_date)

        elif intent == "day":
            current_day = datetime.datetime.now().strftime("%A")
            print("Today is", current_day)

        elif intent == "greeting":
            print("Hello! I'm Arya, how can I help you today?")

        elif intent == "goodbye":
            print("Goodbye! Have an amazing rest of your day!")
            break

        elif intent == "user_doing_well":
            print("That is good to hear.")

        elif intent == "user_feeling_bad":
            print("I hope you feel better soon.")

        elif intent == "thanks":
            print("You're welcome☺️.")

        else:
            print("I'm not sure how to respond to that.")

    except srec.WaitTimeoutError:
        print("Sorry I didn't understand what you said.")

    except srec.UnknownValueError:
        print("Sorry, I didn't understand what you said. Please try again.")
        break

    except srec.RequestError:
        print("Sorry, there was a problem connecting to the speech recognition service.")

    except KeyboardInterrupt:
        print("Goodbye! Have an amazing rest of your day!")
        break
```
