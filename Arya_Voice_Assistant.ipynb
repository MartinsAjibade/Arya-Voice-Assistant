```python
import nltk
from nltk import wordpunct_tokenize
from nltk.stem import PorterStemmer
import pyaudio
import speech_recognition as srec
import datetime
import random
print("Import finished!")
```

```python
# Creating an object
recording = srec.Recognizer()

# Using the computer's microphone as the source of audio
with srec.Microphone() as source:
    print("Arya is listening...")
    audio = recording.listen(source, timeout=5)

print("Audio received!")
```

```python
try: 
    command = recording.recognize_google(audio)
    command = command.lower()
    print("You said:", command)

except:
    print("Sorry I didn't understand what you said.")
```

```python
# Tokenizing the user's command
tokens = wordpunct_tokenize(command)
print("Tokens:", tokens)
```

```python
# Creating a stemming object
stemmer = PorterStemmer()

# Applying stemming to each of the tokens
stemmed_tokens = [stemmer.stem(token) for token in tokens]

print("Stemmed Tokens:", stemmed_tokens)
```

```python
# Identifying the user's intent
if "your" in stemmed_tokens and "name" in stemmed_tokens:
    intent = "arya_name"
elif (
    ("my" in stemmed_tokens and "name" in stemmed_tokens)
    or ("call" in stemmed_tokens and "me" in stemmed_tokens)
    ):
    intent = "user_name"
elif "time" in stemmed_tokens:
    intent = "time"
elif "date" in stemmed_tokens:
    intent = "date"
elif "day" in stemmed_tokens and "date" not in stemmed_tokens:
    intent = "day"
elif "hello" in stemmed_tokens or "hi" in stemmed_tokens or "hey" in stemmed_tokens:
    intent = "greeting"
elif "how" in stemmed_tokens and "you" in stemmed_tokens:
    intent = "how_are_you"
elif "goodbye" in stemmed_tokens or "bye" in stemmed_tokens or "exit" in stemmed_tokens:
    intent = "goodbye"
else:
    intent = "unknown"

print("Intent is:", intent)
```

```python
# Responding based on the user's intent

if intent == "arya_name":
    print("My name is Arya, and you?")

elif intent == "user_name":
    if "my name is" in command:
        name = command.replace("my name is", "").strip()

    elif "call me" in command:
        name = command.replace("call me", "").strip()

    print("Nice to meet you,", name.title())

elif intent == "how_are_you":
    print("I am doing very well, and you?")

elif intent == "time":
    current_time = datetime.datetime.now().strftime("%I:%M %p")
    print("The current time is", current_time)

elif intent == "date":
    current_date = datetime.datetime.now().strftime("%B %d, %Y")
    print("Today's date is", current_date)

elif intent == "day":
    current_day = datetime.datetime.now().strftime("%A")
    print("Today is", current_day)

elif intent == "greeting":
    print("Hello! I'm Arya, how can I help you today?")

elif intent == "goodbye":
    print("Goodbye! Have an amazing rest of your day!")

else:
    print("I'm not sure how to respond to that.")
```
